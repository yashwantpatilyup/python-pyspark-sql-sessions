### Problem Statement

You are given a list of raw email addresses from a customer database.  
Some emails are malformed (no `@`, multiple `@`, spaces, etc.).

Write a function `analyse_emails(emails: list) -> dict` that:
1. Strips and lowercases each email
2. **Validates** each email: must contain exactly **one** `'@'` AND the domain part must contain at least one `'.'`
3. For **valid** emails, count occurrences per domain
4. Returns a dict with:
   - `'domain_counts'`: dict of `{domain: count}` sorted by count descending
   - `'invalid_emails'`: list of the original (after strip+lower) invalid email strings
   - `'total'`: total emails processed
   - `'valid_count'`: count of valid emails
   - `'invalid_count'`: count of invalid emails

---

### Input

```python
emails = [
    "alice@gmail.com",
    "BOB@Yahoo.COM  ",
    "carol AT hotmail DOT com",   # invalid — no @
    "david@gmail.com",
    "eve@@gmail.com",             # invalid — two @
    "frank@outlook.com",
    "GRACE@yahoo.com",
    "henry@gmail.com",
    "irene@gmail",                # invalid — no . in domain
    "jack@hotmail.com",
    " meera@gmail.com ",
    "nina@",                      # invalid — no domain
]
```

---

### Expected Output

```python
{
    'domain_counts': {
        'gmail.com':   4,
        'yahoo.com':   2,
        'hotmail.com': 2,
        'outlook.com': 1,
    },
    'invalid_emails': [
        'carol at hotmail dot com',
        'eve@@gmail.com',
        'irene@gmail',
        'nina@',
    ],
    'total': 12,
    'valid_count': 8,
    'invalid_count': 4,
}
```

---

### Constraints
- `strip()` and `lower()` applied to every email before validation
- Valid = exactly one `@` **AND** at least one `'.'` in the domain part (after the `@`)
- `domain_counts` must be sorted by count descending
- Standard Python only — no `re` module

````python
emails = [
    "alice@gmail.com",
    "BOB@Yahoo.COM  ",
    "carol AT hotmail DOT com",   # invalid — no @
    "david@gmail.com",
    "eve@@gmail.com",             # invalid — two @
    "frank@outlook.com",
    "GRACE@yahoo.com",
    "henry@gmail.com",
    "irene@gmail",                # invalid — no . in domain
    "jack@hotmail.com",
    " meera@gmail.com ",
    "nina@",                      # invalid — no domain
]

clean_emails = [ i.strip().lower() for i in emails]
# print(clean_emails)

valid_emails = []
invalid_emails = []

for i in clean_emails:
    if i.count("@") == 1 and "." in i.split("@")[1] :
        valid_emails.append(i)
    else:
        invalid_emails.append(i)

# print(valid_emails, invalid_emails)

domain_count = {}

for i in valid_emails:
    domain = i.split("@")[1]

    domain_count[domain] = domain_count.get(domain, 0) + 1

sorted_count = []

for domain, count in domain_count.items():
    sorted_count.append((count,domain))

sorted_count.sort(reverse = True)
# print(sorted_count)
sorted_domain_count = {}
for count, domain in sorted_count:
    sorted_domain_count[domain] = count



report ={

    "domain_counts" : sorted_domain_count,
    "invalid emails" : invalid_emails,
    "total_emails": len(emails),
    "valid_emails": len(valid_emails),
    "invalid_emails": len(invalid_emails)

}
print(report)
````
