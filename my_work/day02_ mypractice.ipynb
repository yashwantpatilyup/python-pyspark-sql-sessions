```python
raw = "   data engineer   "

print((raw.strip()))   # 'data engineer'
print(repr(raw.lstrip()))  # 'data engineer   '
print(repr(raw.rstrip()))  # '   data engineer'

# Strip specific characters (not just spaces)
tag = "###IMPORTANT###"
print(tag.strip("#"))   # IMPORTANT
```
