```python
# Problem 2 (Easy)
# Using zip, pair employee names with their salaries and print a report.

names = ["suraj", "yash", "sumit", "raghav", "raman"]
salaries = [1000,2000, 3333, 4928]

for i,j in zip(names,salaries):
    print(f"{i} earns {j}")
```

```python
employees = ["Alice", "Bob", "Carol", "David"]
salaries  = [80000, 95000, 72000, 110000]

print(f"{'Employee':<12} {'Salary':>10}")#this is formattingn in f-string
print("-" * 24)
for name, salary in zip(employees, salaries):
    print(f"{name:<12} ${salary:>9,}")
```

```python
daily_sales = [1200, 3400, 2100, 800, 4500, 1900, 2200]
total = 0

# Problem 3 (Medium)
# Accumulate running total of daily sales; break when it exceeds 10000 — report which day crossed.

for day,i in enumerate(daily_sales, start=1):
    total = total + i

    if total>10000:
        print(f"{day} crossed 10k")
        print(f"total sales is {total}")
        break

```
